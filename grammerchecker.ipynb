```python
def categorize_error(match):
    message = match.message.lower()

    if "pronoun" in message and "third-person" in message:
        return "Subject-Verb Agreement"

    if "subject" in message and "verb" in message:
        return "Subject-Verb Agreement"

    if "spelling" in message or "misspelled" in message:
        return "Spelling"

    if "article" in message or "determiner" in message:
        return "Article/Determiner"

    if "preposition" in message:
        return "Preposition"

    if "tense" in message or "past tense" in message:
        return "Verb Tense"

    if "punctuation" in message:
        return "Punctuation"

    if "capital" in message:
        return "Capitalization"

    return "Other Grammar"
```

```python
def calculate_confidence(match):
    score = 50

    # More than one correction means lower certainty
    if len(match.replacements) == 1:
        score += 30
    elif len(match.replacements) == 2:
        score += 20
    elif len(match.replacements) >= 3:
        score += 10

    # Grammar rules with a clear message
    if match.message:
        score += 10

    # Keep score between 0 and 100
    return min(score, 100)
```

```python
def analyze_text(text):
    matches = tool.check(text)

    results = []

    for match in matches:
        error = text[match.offset:match.offset + match.error_length]
        category = categorize_error(match)
        confidence = calculate_confidence(match)

        results.append({
            "error": error,
            "suggestions": match.replacements[:3],
            "category": category,
            "confidence": confidence,
            "message": match.message,
            "position": match.offset
        })

    corrected_text = language_tool_python.utils.correct(text, matches)

    return {
        "original": text,
        "corrected": corrected_text,
        "errors": results,
        "total_errors": len(results)
    }
```

```python
import re

def normalize_text(text):
    text = text.lower()
    text = re.sub(r'[^\w\s]', '', text)
    text = re.sub(r'\s+', ' ', text)
    return text.strip()


def word_accuracy(original, predicted, expected):
    original_words = set(normalize_text(original).split())
    predicted_words = set(normalize_text(predicted).split())
    expected_words = set(normalize_text(expected).split())

    # Words that the reference correction added
    needed_words = expected_words - original_words

    # Words that LanguageTool added
    predicted_added = predicted_words - original_words

    if len(needed_words) == 0:
        return 1.0

    correctly_added = needed_words.intersection(predicted_added)

    return len(correctly_added) / len(needed_words)
```

```python
def correction_success(original, predicted):
    original_n = normalize_text(original)
    predicted_n = normalize_text(predicted)

    # If nothing changed
    if original_n == predicted_n:
        return 0

    return 1


results_df["Correction Made"] = results_df.apply(
    lambda row: correction_success(
        row["Original"],
        row["Predicted"]
    ),
    axis=1
)

print(
    "Sentences where LanguageTool made a correction:",
    results_df["Correction Made"].sum()
)

print(
    "Correction rate:",
    results_df["Correction Made"].mean() * 100,
    "%"
)
```

```python
import language_tool_python
import spacy

# Load spaCy
nlp = spacy.load("en_core_web_sm")

# Load LanguageTool
tool = language_tool_python.LanguageTool("en-US")


# -----------------------------
# Error Category
# -----------------------------
def categorize_error(match):
    message = match.message.lower()

    if "pronoun" in message and "third-person" in message:
        return "Subject-Verb Agreement"

    if "subject" in message and "verb" in message:
        return "Subject-Verb Agreement"

    if "spelling" in message or "misspelled" in message:
        return "Spelling"

    if "article" in message or "determiner" in message:
        return "Article/Determiner"

    if "preposition" in message:
        return "Preposition"

    if "tense" in message or "past tense" in message:
        return "Verb Tense"

    if "punctuation" in message:
        return "Punctuation"

    if "capital" in message:
        return "Capitalization"

    return "Other Grammar"


# -----------------------------
# Confidence Score
# -----------------------------
def calculate_confidence(match):
    score = 50

    if len(match.replacements) == 1:
        score += 30
    elif len(match.replacements) == 2:
        score += 20
    elif len(match.replacements) >= 3:
        score += 10

    if match.message:
        score += 10

    return min(score, 100)


# -----------------------------
# Combined NLP Analysis
# -----------------------------
def analyze_with_nlp(text):

    # 1. spaCy analysis
    doc = nlp(text)

    tokens = []

    for token in doc:
        tokens.append({
            "word": token.text,
            "pos": token.pos_,
            "lemma": token.lemma_
        })

    # 2. LanguageTool analysis
    matches = tool.check(text)

    errors = []

    for match in matches:

        error = text[
            match.offset:
            match.offset + match.error_length
        ]

        errors.append({
            "error": error,
            "suggestions": match.replacements[:3],
            "category": categorize_error(match),
            "confidence": calculate_confidence(match),
            "message": match.message,
            "position": match.offset
        })

    # 3. Correct sentence
    corrected_text = language_tool_python.utils.correct(
        text,
        matches
    )

    # 4. Return result
    return {
        "original": text,
        "corrected": corrected_text,
        "tokens": tokens,
        "errors": errors,
        "total_errors": len(errors)
    }

print("Combined NLP function created successfully!")
```

```python
def categorize_error(match):
    message = match.message.lower()
    rule = match.rule_id.lower() if hasattr(match, "rule_id") else ""

    # Punctuation
    if (
        "comma" in message
        or "full stop" in message
        or "punctuation" in message
        or "punctuation" in rule
    ):
        return "Punctuation"

    # Spelling
    if (
        "spelling mistake" in message
        or "misspelled" in message
        or "possible spelling" in message
    ):
        return "Spelling"

    # Subject-Verb Agreement
    if (
        "subject" in message and "verb" in message
    ) or (
        "third-person" in message
    ):
        return "Subject-Verb Agreement"

    # Articles / Determiners
    if (
        "article" in message
        or "determiner" in message
    ):
        return "Article/Determiner"

    # Prepositions
    if "preposition" in message:
        return "Preposition"

    # Verb tense
    if (
        "tense" in message
        or "past tense" in message
        or "present tense" in message
    ):
        return "Verb Tense"

    # Capitalization
    if (
        "capital" in message
        or "uppercase" in message
    ):
        return "Capitalization"

    return "Other Grammar"
```

```python
import re
from difflib import SequenceMatcher
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# Tokenize words + punctuation
TOKEN_PATTERN = re.compile(r"\w+|[^\w\s]", re.UNICODE)

def get_tokens_with_spans(text):
    tokens = []

    for match in TOKEN_PATTERN.finditer(text):
        tokens.append({
            "token": match.group(),
            "start": match.start(),
            "end": match.end()
        })

    return tokens


def normalize_token(token):
    return token.lower()


def get_reference_error_tokens(source, target):
    """
    Finds source tokens that differ from the reference correction.
    These are treated as reference error tokens.
    """

    source_tokens = [
        normalize_token(x["token"])
        for x in get_tokens_with_spans(source)
    ]

    target_tokens = [
        normalize_token(x["token"])
        for x in get_tokens_with_spans(target)
    ]

    matcher = SequenceMatcher(
        None,
        source_tokens,
        target_tokens
    )

    error_indices = set()

    for tag, i1, i2, j1, j2 in matcher.get_opcodes():

        if tag != "equal":
            for i in range(i1, i2):
                error_indices.add(i)

    return error_indices


def get_predicted_error_tokens(source, matches):
    """
    Finds source tokens covered by LanguageTool errors.
    """

    tokens = get_tokens_with_spans(source)

    predicted_indices = set()

    for match in matches:

        error_start = match.offset
        error_end = match.offset + match.error_length

        for i, token in enumerate(tokens):

            token_start = token["start"]
            token_end = token["end"]

            # Check whether token overlaps LanguageTool error span
            if (
                token_start < error_end
                and token_end > error_start
            ):
                predicted_indices.add(i)

    return predicted_indices


# Store all true/predicted labels
y_true = []
y_pred = []

for _, row in test_df.iterrows():

    source = row["source"]
    target = row["target"]

    # Source tokens
    tokens = get_tokens_with_spans(source)

    # Reference error tokens
    reference_errors = get_reference_error_tokens(
        source,
        target
    )

    # LanguageTool predictions
    matches = tool.check(source)

    predicted_errors = get_predicted_error_tokens(
        source,
        matches
    )

    # Create labels for every source token
    for i in range(len(tokens)):

        if i in reference_errors:
            y_true.append(1)
        else:
            y_true.append(0)

        if i in predicted_errors:
            y_pred.append(1)
        else:
            y_pred.append(0)


# Calculate metrics
accuracy = accuracy_score(y_true, y_pred)
precision = precision_score(
    y_true,
    y_pred,
    zero_division=0
)
recall = recall_score(
    y_true,
    y_pred,
    zero_division=0
)
f1 = f1_score(
    y_true,
    y_pred,
    zero_division=0
)


print("===== ERROR DETECTION EVALUATION =====")
print()
print("Total tokens:", len(y_true))
print()
print("Accuracy :", round(accuracy * 100, 2), "%")
print("Precision:", round(precision * 100, 2), "%")
print("Recall   :", round(recall * 100, 2), "%")
print("F1-Score :", round(f1 * 100, 2), "%")
```

```python
# ==========================================
# STEP 15: FINAL GRAMMAR CHECKER
# ==========================================

def grammar_checker(text):

    # 1. NLP processing
    doc = nlp(text)

    # 2. LanguageTool error detection
    matches = tool.check(text)

    errors = []

    # 3. Extract information about each error
    for match in matches:

        error_text = text[
            match.offset:
            match.offset + match.error_length
        ]

        suggestions = match.replacements[:3]

        category = categorize_error(match)

        confidence = calculate_confidence(match)

        errors.append({
            "error": error_text,
            "suggestions": suggestions,
            "category": category,
            "confidence": confidence,
            "message": match.message,
            "position": match.offset
        })

    # 4. Generate corrected sentence
    corrected_text = language_tool_python.utils.correct(
        text,
        matches
    )

    # 5. Return final result
    return {
        "original": text,
        "corrected": corrected_text,
        "total_errors": len(errors),
        "errors": errors
    }


print("Final grammar checker is ready!")
```

```python
# ==========================================
# RESTORE NLP ENVIRONMENT
# ==========================================

import language_tool_python
import spacy

# Load spaCy
nlp = spacy.load("en_core_web_sm")

# Load LanguageTool
tool = language_tool_python.LanguageTool("en-US")

print("spaCy loaded:", nlp is not None)
print("LanguageTool loaded:", tool is not None)
```

```python
# ==========================================
# HELPER FUNCTIONS
# ==========================================

def categorize_error(match):

    message = match.message.lower()
    rule = match.rule_id.lower() if hasattr(match, "rule_id") else ""

    # Punctuation
    if (
        "comma" in message
        or "full stop" in message
        or "punctuation" in message
        or "punctuation" in rule
    ):
        return "Punctuation"

    # Spelling
    if (
        "spelling mistake" in message
        or "misspelled" in message
        or "possible spelling" in message
    ):
        return "Spelling"

    # Subject-Verb Agreement
    if (
        ("subject" in message and "verb" in message)
        or "third-person" in message
    ):
        return "Subject-Verb Agreement"

    # Articles / Determiners
    if (
        "article" in message
        or "determiner" in message
    ):
        return "Article/Determiner"

    # Prepositions
    if "preposition" in message:
        return "Preposition"

    # Verb tense
    if (
        "tense" in message
        or "past tense" in message
        or "present tense" in message
    ):
        return "Verb Tense"

    # Capitalization
    if (
        "capital" in message
        or "uppercase" in message
    ):
        return "Capitalization"

    return "Other Grammar"


def calculate_confidence(match):

    score = 50

    if len(match.replacements) == 1:
        score += 30

    elif len(match.replacements) == 2:
        score += 20

    elif len(match.replacements) >= 3:
        score += 10

    if match.message:
        score += 10

    return min(score, 100)


print("Helper functions loaded!")
```

```python
# ==========================================
# FINAL GRAMMAR CHECKER
# ==========================================

def grammar_checker(text):

    matches = tool.check(text)

    errors = []

    for match in matches:

        error_text = text[
            match.offset:
            match.offset + match.error_length
        ]

        errors.append({
            "error": error_text,
            "suggestions": match.replacements[:3],
            "category": categorize_error(match),
            "confidence": calculate_confidence(match),
            "message": match.message,
            "position": match.offset
        })

    corrected_text = language_tool_python.utils.correct(
        text,
        matches
    )

    return {
        "original": text,
        "corrected": corrected_text,
        "total_errors": len(errors),
        "errors": errors
    }


print("Grammar checker loaded!")
```
